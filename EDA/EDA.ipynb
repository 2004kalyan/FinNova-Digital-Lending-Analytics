```python
import pandas as pd
applications = pd.read_csv("applications_raw.csv")
loans = pd.read_csv("loans_raw.csv")

# 1. Total applications
total_applications = applications["application_id"].nunique()

# 2. Approved applications
approved_applications = (
    applications[applications["application_status"] == "Approved"]
    ["application_id"].nunique()
)

# 3. Disbursed loans
disbursed_loans = loans["loan_id"].nunique()

# 4. Approval rate
approval_rate = approved_applications / total_applications * 100

# 5. Approval-to-disbursement rate
approval_to_disbursement = disbursed_loans / approved_applications * 100

# 6. Application-to-disbursement rate
application_to_disbursement = disbursed_loans / total_applications * 100

print("Total Applications:", total_applications)
print("Approved Applications:", approved_applications)
print("Disbursed Loans:", disbursed_loans)
print("Approval Rate:", round(approval_rate, 2), "%")
print("Approval → Disbursement:", round(approval_to_disbursement, 2), "%")
print("Application → Disbursement:", round(application_to_disbursement, 2), "%")
```

```python
total_requested = applications["requested_amount"].sum()

total_approved = applications["approved_amount"].sum()

total_disbursed = loans["disbursement_amount"].sum()

avg_requested = applications["requested_amount"].mean()

avg_disbursed = loans["disbursement_amount"].mean()

print("Total Requested: ₹", round(total_requested, 2))
print("Total Approved: ₹", round(total_approved, 2))
print("Total Disbursed: ₹", round(total_disbursed, 2))
print("Average Requested: ₹", round(avg_requested, 2))
print("Average Disbursed: ₹", round(avg_disbursed, 2))
```

```python
import pandas as pd

# Load datasets
applications = pd.read_csv("applications_raw.csv")
loans = pd.read_csv("loans_raw.csv")
credit = pd.read_csv("credit_assessments_raw.csv")
repayments = pd.read_csv("repayments_raw.csv")

# Create risk bands
bins = [299, 549, 649, 699, 749, 850]
labels = [
    "High Risk",
    "Moderate Risk",
    "Medium Risk",
    "Low Risk",
    "Very Low Risk"
]

credit["risk_band"] = pd.cut(
    credit["credit_score"],
    bins=bins,
    labels=labels
)

# Remove duplicate loan/application records
loans_unique = loans.drop_duplicates("loan_id")
applications_unique = applications.drop_duplicates("application_id")

# Connect loans with risk bands
loan_risk = (
    loans_unique
    .merge(
        applications_unique[["application_id", "customer_id"]],
        on="application_id",
        how="left"
    )
    .merge(
        credit[["application_id", "risk_band"]],
        on="application_id",
        how="left"
    )
)

# Determine whether each loan ever reached default
loan_default = (
    repayments.groupby("loan_id")["default_flag"]
    .max()
    .reset_index()
)

# Add default information
loan_risk = loan_risk.merge(
    loan_default,
    on="loan_id",
    how="left"
)

# Calculate default rate by risk band
default_by_risk = (
    loan_risk
    .dropna(subset=["risk_band"])
    .groupby("risk_band", observed=False)
    .agg(
        loans=("loan_id", "nunique"),
        defaulted_loans=("default_flag", "sum")
    )
)

default_by_risk["default_rate"] = (
    default_by_risk["defaulted_loans"]
    / default_by_risk["loans"] * 100
)

default_by_risk
```

```python
customers = pd.read_csv("customers_raw.csv")
loan_customer = (
    loans_unique
    .merge(
        applications_unique[["application_id", "customer_id"]],
        on="application_id",
        how="left"
    )
    .merge(
        customers[["customer_id", "customer_type"]],
        on="customer_id",
        how="left"
    )
    .merge(
        loan_default,
        on="loan_id",
        how="left"
    )
)

default_by_customer_type = (
    loan_customer
    .groupby("customer_type")
    .agg(
        loans=("loan_id", "nunique"),
        defaulted_loans=("default_flag", "sum")
    )
)

default_by_customer_type["default_rate"] = (
    default_by_customer_type["defaulted_loans"]
    / default_by_customer_type["loans"] * 100
)

default_by_customer_type
```

```python
default_by_employment = (
    loan_customer
    .merge(
        customers[["customer_id", "employment_type"]],
        on="customer_id",
        how="left"
    )
    .groupby("employment_type")
    .agg(
        loans=("loan_id", "nunique"),
        defaulted_loans=("default_flag", "sum")
    )
)

default_by_employment["default_rate"] = (
    default_by_employment["defaulted_loans"]
    / default_by_employment["loans"] * 100
)

default_by_employment.sort_values("default_rate", ascending=False)
```

```python
loan_income = (
    loan_customer
    .merge(
        customers[["customer_id", "monthly_income"]],
        on="customer_id",
        how="left"
    )
)

loan_income["income_band"] = pd.cut(
    loan_income["monthly_income"],
    bins=[0, 30000, 50000, 75000, 100000, float("inf")],
    labels=[
        "Below 30K",
        "30K-50K",
        "50K-75K",
        "75K-100K",
        "100K+"
    ]
)

default_by_income = (
    loan_income
    .dropna(subset=["income_band"])
    .groupby("income_band", observed=False)
    .agg(
        loans=("loan_id", "nunique"),
        defaulted_loans=("default_flag", "sum")
    )
)

default_by_income["default_rate"] = (
    default_by_income["defaulted_loans"]
    / default_by_income["loans"] * 100
)

default_by_income
```

```python
monthly_disbursement = (
    loans
    .groupby(loans["disbursement_date"].dt.to_period("M"))
    .agg(
        loans=("loan_id", "nunique"),
        disbursement_amount=("disbursement_amount", "sum")
    )
    .reset_index()
)

monthly_disbursement["month"] = (
    monthly_disbursement["disbursement_date"]
    .astype(str)
)

monthly_disbursement
```
